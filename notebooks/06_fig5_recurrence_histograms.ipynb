# 06 — Figure 5: flare recurrence-time histograms

Reproduces **Figure 5** of the paper.

Recurrence time is defined as the **peak-to-peak interval** between two successive flares of class C or above produced by the same active region (Equation 1 of the paper). For each AR we draw a histogram of the 78 (AR 13663) and 81 (AR 13664) intervals.

Per the published statistics:

| AR | N flares | N intervals | Mean (h) | Median (h) | Notes |
|---|---:|---:|---:|---:|---|
| 13663 | 79 | 78 | 1.766 (filtered, RT < 10 h) | 1.2 | one outlier (≥ 10 h) excluded |
| 13664 | 82 | 81 | 3.591 (all data)            | 1.6 | full distribution, including 5 long-tail outliers |

Inputs:
* `data/dados_analises_may_2024_flares.xlsx`, sheets `time_flare_interval_AR13663` and `time_flare_interval_AR13664`.

Outputs:
* `figures/grafico_recurrence_time_AR13663_novo.png` (Fig. 5a)
* `figures/grafico_recurrence_time_AR13664_novo.png` (Fig. 5b)


```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

DATA_DIR = Path('..').resolve() / 'data'
FIG_DIR  = Path('..').resolve() / 'figures'
FIG_DIR.mkdir(parents=True, exist_ok=True)

XLSX = DATA_DIR / 'dados_analises_may_2024_flares.xlsx'
print('Sheets available:', pd.ExcelFile(XLSX).sheet_names)
```

```python
def plot_recurrence(sheet_name, ar_label, total_intervals,
                    mean_hours, median_hours, outfile):
    """Histogram of the recurrence-time distribution for a single AR.

    Parameters
    ----------
    sheet_name : str
        Name of the Excel sheet that contains 'Recurrence Time' and
        'Quantities' columns (the latter is the count per 1-hour bin).
    ar_label : str
        Label drawn in the title (e.g. 'AR 13663').
    total_intervals : int
        Number of recurrence intervals (= N_flares − 1).
    mean_hours, median_hours : float
        Statistics shown as red/green dashed vertical lines.
    outfile : Path
        Where to save the PNG.
    """
    df = pd.read_excel(XLSX, sheet_name=sheet_name)
    rt  = df['Recurrence Time']
    qty = df['Quantities']

    # Shift bars so each bar covers [t-1, t] hours (paper convention)
    adjusted = rt - 1

    plt.figure(figsize=(10, 6))
    plt.bar(adjusted, qty, width=1.0, color='orange',
            edgecolor='black', align='edge')

    plt.axvline(x=mean_hours,   color='red',   linestyle='--',
                linewidth=2, label=f'Mean {mean_hours:.3f} hrs')
    plt.axvline(x=median_hours, color='green', linestyle='--',
                linewidth=2, label=f'Median {median_hours:.1f} hrs')

    plt.xlim(left=0)
    plt.xticks(range(0, int(max(rt)) + 2, 2))
    plt.ylim(0, 40)

    plt.title(f'{ar_label}\nTotal Recurrence Intervals: {total_intervals}',
              fontsize=16)
    plt.xlabel('Flare Recurrence Time (hrs)', fontsize=16)
    plt.ylabel('Number of Flares', fontsize=16)
    plt.legend()

    plt.savefig(outfile, dpi=300, bbox_inches='tight')
    plt.show()
```

## Figure 5a — AR 13663


```python
plot_recurrence(
    sheet_name='time_flare_interval_AR13663',
    ar_label='AR 13663',
    total_intervals=78,
    mean_hours   = 1 + 45/60 + 57/3600,   # 1.766 h (filtered, RT < 10 h)
    median_hours = 1 + 12/60,             # 1.2 h
    outfile = FIG_DIR / 'grafico_recurrence_time_AR13663_novo.png',
)
```

## Figure 5b — AR 13664


```python
plot_recurrence(
    sheet_name='time_flare_interval_AR13664',
    ar_label='AR 13664',
    total_intervals=81,
    mean_hours   = 3 + 35/60 + 28/3600,   # 3.591 h (full set, including long-tail outliers)
    median_hours = 1 + 37/60,             # 1.6 h
    outfile = FIG_DIR / 'grafico_recurrence_time_AR13664_novo.png',
)
```
