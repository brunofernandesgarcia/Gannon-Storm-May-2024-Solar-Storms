# 03 — Figure 1: GOES/XRS 1–8 Å soft X-ray flux (May 01–14, 2024)

Reproduces **Figure 1** of the paper.

Input data: `data/goes_xray_flux_may2024.csv` (NOAA SWPC 1-minute GOES Longwave 1–8 Å series, exported as semicolon-separated CSV with comma as decimal point).

Output: `figures/flux_x_ray_may2024_revised.png`.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

DATA = Path('..').resolve() / 'data' / 'goes_xray_flux_may2024.csv'
FIG_DIR = Path('..').resolve() / 'figures'
FIG_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# The CSV exported from NOAA SWPC uses ';' as separator and ',' as decimal
df = pd.read_csv(DATA, sep=';', decimal=',')
df['time'] = pd.to_datetime(df.iloc[:, 0])
df['flux'] = df.iloc[:, 1]
df.head()
```

```python
fig, ax = plt.subplots(figsize=(14, 7), dpi=100)

ax.plot(df['time'], df['flux'], color='#2563EB', lw=1.5, label='GOES 1–8 Å')

# Log scale + canonical C/M/X thresholds
ax.set_yscale('log')
ax.set_ylim(1e-7, 1e-3)

for letra, valor in [('C', 1e-6), ('M', 1e-5), ('X', 1e-4)]:
    ax.axhline(y=valor, color='gray', linestyle='--', alpha=0.3, linewidth=0.8)
    ax.text(df['time'].max(), valor, f' {letra}',
            verticalalignment='bottom', fontsize=12, color='gray', fontweight='bold')

ax.set_title('GOES 1–8 Å X-ray Flux', color='blue', fontsize=16, pad=20)
ax.set_ylabel('Watts / m²', fontsize=12)
ax.set_xlabel('Time (UT)', fontsize=12)
ax.xaxis.set_major_formatter(mdates.DateFormatter('%d/%b\n%Hh'))
plt.xticks(rotation=0)

ax.grid(True, which='major', color='gray', linestyle='-', alpha=0.1)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig(FIG_DIR / 'flux_x_ray_may2024_revised.png', dpi=300, bbox_inches='tight')
plt.show()
```
